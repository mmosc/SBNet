```python
import pandas as pd
import os
```

```python
data_folder = '/home/marta/jku/2025_ws_mmsr/data/'
full_data_folder = f'{data_folder}/full'
retrieval_data_folder = f'{data_folder}/retrieval'
binary_classification_data_folder = f'{data_folder}/binary_classification'
```

```python
def filter_dataset(input_file_path: str, output_file_path: str, ids: set):
    track_genres = pd.read_csv(input_file_path, sep='\t')
    filtered_tracks = track_genres[track_genres['id'].isin(ids)]
    filtered_tracks.to_csv(output_file_path, sep='\t', index=False)
```

```python
binary_classification_genres = pd.read_csv(f'{binary_classification_data_folder}/binary_full.tsv', sep='\t')
```

```python
full_genres = pd.read_csv(f'{full_data_folder}/id_genres_mmsr.tsv', sep='\t')
```

```python
full_ids = set(full_genres['id'].tolist())
binary_classification_ids = set(binary_classification_genres['i'].tolist())
retrieval_ids = full_ids - binary_classification_ids
```

```python
len(full_ids), len(binary_classification_ids), len(retrieval_ids)
```

```python
directory = os.fsencode(full_data_folder)

for file in os.listdir(directory):
    filename = os.fsdecode(file)
    if filename.endswith(".tsv"):
        print(f'Filtering {filename}...')
        input_file_path = os.path.join(full_data_folder, filename)
        output_file_path = os.path.join(retrieval_data_folder, filename)
        filter_dataset(input_file_path, output_file_path, retrieval_ids)
        continue
    else:
        continue
```

```python
directory = os.fsencode(full_data_folder)

for file in os.listdir(directory):
    filename = os.fsdecode(file)
    if filename.endswith(".tsv"):
        print(f'Filtering {filename}...')
        input_file_path = os.path.join(full_data_folder, filename)
        output_file_path = os.path.join(binary_classification_data_folder, filename)
        filter_dataset(input_file_path, output_file_path, binary_classification_ids)
        continue
    else:
        continue
```
